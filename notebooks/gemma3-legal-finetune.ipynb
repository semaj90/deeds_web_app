## GOLDEN RULES

```
1. ALWAYS save BEFORE calling for_inference()
   for_inference() activates torch.compile / FakeTensorMode.
   ONE-WAY DOOR — cannot save after. Even del model won't fix it.
   Only a full runtime restart clears it.

   ORDER: Load → Merge LoRA → Save ALL exports → THEN for_inference()

2. MUST call merge_and_unload() BEFORE merged/GGUF saves
   PeftModel.from_pretrained() wraps the model, but Unsloth's
   save_pretrained_merged() and save_pretrained_gguf() don't
   detect it. They save the BASE model without your training.
   Call model = model.merge_and_unload() first to fuse LoRA
   weights permanently into the base model.
   
   WARNING SIGN: "Model is not a PeftModel (no Lora adapters detected)"
   = your save does NOT contain legal fine-tuning!

3. Use google/gemma-3-12b-it (NOT unsloth pre-quantized) for reload
   The unsloth/gemma-3-12b-it-bnb-4bit version can have quant_state
   mismatches with newer bitsandbytes versions.

4. Chat template: content must be list-of-dicts, NOT plain string
   {"role": "user", "content": [{"type": "text", "text": prompt}]}

5. TRT-LLM engines built on A100 (SM 80) run on RTX 3060 Ti (SM 86)
   Both are Ampere. SM 80 PTX forward-compiles to SM 86 via JIT.

6. GGUF save needs config.json — save to LOCAL disk first, copy to Drive
   save_pretrained_gguf() to an empty Drive folder fails with
   "config.json does not exist". Always save locally, then copy.
```

Uses llama.cpp converter directly on files — no Python model loading.

````python
# ============================================
# GGUF via llama.cpp direct — no Python model loading
# Reads safetensors + config.json directly from disk
# For Ollama FALLBACK (TRT-LLM is primary)
# ============================================
import subprocess, shutil
from pathlib import Path

DRIVE_BASE = Path("/content/drive/MyDrive")
MERGED_DIR = DRIVE_BASE / "gemma3-legal-merged-16bit"
GGUF_DIR = DRIVE_BASE / "gemma3-legal-gguf"
GGUF_DIR.mkdir(parents=True, exist_ok=True)

# Step 1: Ensure llama.cpp is available
LLAMA_CPP = Path("/root/.unsloth/llama.cpp")
if not LLAMA_CPP.exists():
    print("Cloning llama.cpp...")
    subprocess.run([
        "git", "clone", "--depth=1",
        "https://github.com/ggerganov/llama.cpp.git",
        str(LLAMA_CPP),
    ], check=True)

# Step 2: Convert merged HF → BF16 GGUF (~3 min)
BF16_GGUF = Path("gemma3-legal.BF16.gguf")
print("Converting HF → BF16 GGUF (~3 min)...")
subprocess.run([
    "python3", str(LLAMA_CPP / "convert_hf_to_gguf.py"),
    str(MERGED_DIR),
    "--outfile", str(BF16_GGUF),
    "--outtype", "bf16",
], check=True)
print(f"BF16 GGUF: {BF16_GGUF.stat().st_size / 1e9:.1f} GB")

# Step 3: Build llama-quantize if needed
QUANTIZE = LLAMA_CPP / "build" / "bin" / "llama-quantize"
if not QUANTIZE.exists():
    print("Building llama-quantize...")
    build_dir = LLAMA_CPP / "build"
    build_dir.mkdir(exist_ok=True)
    subprocess.run(["cmake", "..", "-DCMAKE_BUILD_TYPE=Release"],
                   cwd=str(build_dir), check=True)
    subprocess.run(["cmake", "--build", ".", "--target", "llama-quantize", "-j4"],
                   cwd=str(build_dir), check=True)

# Step 4: Quantize BF16 → Q4_K_M (~10 min)
Q4_GGUF = Path("gemma3-legal.Q4_K_M.gguf")
print("Quantizing BF16 → Q4_K_M (~10 min)...")
subprocess.run([str(QUANTIZE), str(BF16_GGUF), str(Q4_GGUF), "Q4_K_M"], check=True)
print(f"Q4_K_M GGUF: {Q4_GGUF.stat().st_size / 1e9:.1f} GB")

# Step 5: Copy to Drive + cleanup
shutil.copy2(str(Q4_GGUF), str(GGUF_DIR / Q4_GGUF.name))
BF16_GGUF.unlink(missing_ok=True)  # Free ~24 GB disk
print(f"\nGGUF saved to {GGUF_DIR / Q4_GGUF.name}")
print("Download this for Ollama fallback deployment.")
````

### 4E. Copy Merged Shards to Google Drive (resumable)

Re-run safe — already-copied shards are skipped by size check.

````python
LOCAL_MERGED = Path("gemma3-legal-merged-16bit")
if not LOCAL_MERGED.exists():
    print("No merged model found locally. Run 4C first.")
else:
    MERGED_DIR.mkdir(parents=True, exist_ok=True)
    all_files = sorted(LOCAL_MERGED.iterdir())

    print(f"Copying {len(all_files)} files to Drive (skips existing)...\n")
    copied, skipped = 0, 0

    for i, src in enumerate(all_files, 1):
        dst = MERGED_DIR / src.name
        src_size = src.stat().st_size
        size_gb = src_size / (1024**3)

        if dst.exists() and dst.stat().st_size == src_size:
            print(f"  [{i}/{len(all_files)}] SKIP {src.name} ({size_gb:.2f} GB)")
            skipped += 1
            continue

        print(f"  [{i}/{len(all_files)}] {src.name} ({size_gb:.2f} GB)...", end=" ", flush=True)
        t0 = time.time()
        shutil.copy2(str(src), str(dst))
        elapsed = time.time() - t0
        speed = src_size / elapsed / (1024**2) if elapsed > 0 else 0
        print(f"done ({elapsed:.0f}s, {speed:.0f} MB/s)")
        copied += 1

    print(f"\nDone: {copied} copied, {skipped} skipped")
    print(f"If Colab disconnects mid-copy, just re-run this cell.")
````

````python
from transformers import TextStreamer

def ask(prompt, system="You are a legal AI assistant specialized in evidence analysis, case law, and legal document processing.", max_tokens=512):
    """Send a properly-formatted chat prompt to the model.
    
    IMPORTANT: content must be list-of-dicts format for vision models.
    [{"type": "text", "text": prompt}] NOT just a plain string.
    Plain string causes: TypeError: string indices must be integers
    """
    messages = [
        {"role": "system", "content": [{"type": "text", "text": system}]},
        {"role": "user", "content": [{"type": "text", "text": prompt}]},
    ]

    input_text = tokenizer.apply_chat_template(
        messages, add_generation_prompt=True, tokenize=False,
    )
    inputs = tokenizer(input_text, return_tensors="pt").to(model.device)

    streamer = TextStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True)

    print(f"\n{'='*60}")
    print(f"Prompt: {prompt}")
    print(f"{'='*60}")

    _ = model.generate(
        **inputs,
        streamer=streamer,
        max_new_tokens=max_tokens,
        temperature=0.3,
        top_p=0.9,
        do_sample=True,
        use_cache=True,
    )
    print()
````

### PTX Cross-Architecture Portability

~30-60s first load, ~10% perf penalty vs native build.

### 7A. Extract Text-Only Model from Vision Model

````python
# ============================================
# 7A: Extract text-only model from vision model
# Remaps weight keys, drops vision tower + projector
# Required before AWQ — prevents architecture mismatch
# ============================================
import json, shutil
from pathlib import Path
from safetensors.torch import load_file, save_file

DRIVE_BASE = Path("/content/drive/MyDrive")
MERGED_DIR = DRIVE_BASE / "gemma3-legal-merged-16bit"
TEXT_DIR = Path("gemma3-legal-text-only")
TEXT_DIR.mkdir(exist_ok=True)

# 1. Build text-only config from inner text_config
config = json.loads((MERGED_DIR / "config.json").read_text())
text_config = config.get("text_config", {})
text_only = {**text_config}
text_only["architectures"] = ["Gemma3ForCausalLM"]
text_only["model_type"] = "gemma3"
text_only["torch_dtype"] = config.get("torch_dtype", "bfloat16")
text_only["bos_token_id"] = config.get("bos_token_id", 2)
text_only["eos_token_id"] = config.get("eos_token_id", 1)
text_only["pad_token_id"] = config.get("pad_token_id", 0)
(TEXT_DIR / "config.json").write_text(json.dumps(text_only, indent=2))
print(f"Config: Gemma3ForCausalLM")
print(f"  hidden_size: {text_only.get('hidden_size')}")
print(f"  num_layers: {text_only.get('num_hidden_layers')}")
print(f"  sliding_window: {text_only.get('sliding_window')}")

# 2. Copy tokenizer files
for f in ["tokenizer.json", "tokenizer_config.json", "tokenizer.model",
          "special_tokens_map.json", "generation_config.json"]:
    src = MERGED_DIR / f
    if src.exists():
        shutil.copy2(str(src), str(TEXT_DIR / f))

# 3. Remap weights — strip language_model prefix, drop vision
for sf in sorted(MERGED_DIR.glob("*.safetensors")):
    print(f"\n{sf.name} ({sf.stat().st_size / 1e9:.2f} GB)...")
    tensors = load_file(str(sf))
    new_tensors = {}
    skipped = 0

    for key, value in tensors.items():
        if "vision_tower" in key or "multi_modal_projector" in key:
            skipped += 1
            continue
        if key.startswith("model.language_model."):
            new_key = "model." + key[len("model.language_model."):]
        elif key.startswith("language_model."):
            new_key = key[len("language_model."):]
        else:
            new_key = key
        new_tensors[new_key] = value

    save_file(new_tensors, str(TEXT_DIR / sf.name))
    print(f"  Kept: {len(new_tensors)} | Dropped vision: {skipped}")

total = sum(f.stat().st_size for f in TEXT_DIR.iterdir() if f.is_file()) / 1e9
print(f"\nText-only model: {total:.1f} GB (vision tower removed)")
print("Ready for AWQ quantization (Cell 7B).")
````

### 7B. AWQ INT4 Quantization (~30 min)

Quantizes the **text-only** model from 7A.
Uses AutoAWQ (not Unsloth) — clean HuggingFace load, no vision mismatch.

````python
# ============================================
# 7B: AWQ INT4 quantization for TRT-LLM engine
# Input: text-only model from 7A (~20 GB)
# Output: AWQ INT4 checkpoint (~6 GB)
# ============================================
!pip install autoawq safetensors

from awq import AutoAWQForCausalLM
from transformers import AutoTokenizer
from pathlib import Path
import time

DRIVE_BASE = Path("/content/drive/MyDrive")
TEXT_DIR = Path("gemma3-legal-text-only")
AWQ_DIR = DRIVE_BASE / "gemma3-legal-awq-int4"

assert (TEXT_DIR / "config.json").exists(), "Run Cell 7A first!"

print("Loading text-only model for AWQ quantization...")
t0 = time.time()
model_awq = AutoAWQForCausalLM.from_pretrained(
    str(TEXT_DIR), device_map="auto", torch_dtype="auto",
)
tok_awq = AutoTokenizer.from_pretrained(str(TEXT_DIR))
print(f"Model loaded in {time.time() - t0:.0f}s")

quant_config = {
    "zero_point": True,
    "q_group_size": 128,
    "w_bit": 4,
    "version": "GEMM",
}

print("Quantizing to INT4 AWQ (~30 min)...")
t0 = time.time()
model_awq.quantize(tok_awq, quant_config=quant_config)
print(f"Quantization done in {time.time() - t0:.0f}s")

AWQ_DIR.mkdir(parents=True, exist_ok=True)
model_awq.save_quantized(str(AWQ_DIR))
tok_awq.save_pretrained(str(AWQ_DIR))

total = sum(f.stat().st_size for f in AWQ_DIR.rglob("*") if f.is_file()) / 1e9
print(f"\nAWQ INT4 saved: {AWQ_DIR} ({total:.1f} GB)")
print("Ready for TRT-LLM engine build (Cell 7C).")
````

### 7D. Export SigLIP Vision Tower as ONNX

Exports the vision encoder separately for VLM analysis (POI photos).
Can be converted to TensorRT later with `trtexec --onnx=... --fp16`.

````python
# ============================================
# 7D: Export SigLIP vision tower as ONNX
# For VLM analysis (POI photos, evidence images)
# Separate from text engine — loaded alongside in Triton
# ============================================
import torch
from transformers import AutoModel
from pathlib import Path

DRIVE_BASE = Path("/content/drive/MyDrive")
MERGED_DIR = DRIVE_BASE / "gemma3-legal-merged-16bit"
VISION_DIR = DRIVE_BASE / "gemma3-legal-vision-tower"
VISION_DIR.mkdir(parents=True, exist_ok=True)

# Load FULL vision model to extract SigLIP
print("Loading vision model to extract SigLIP...")
full_model = AutoModel.from_pretrained(
    str(MERGED_DIR), torch_dtype=torch.float16, device_map="cpu",
)

vision_tower = full_model.vision_tower
projector = full_model.multi_modal_projector

# Export vision tower as ONNX
dummy_pixel = torch.randn(1, 3, 896, 896, dtype=torch.float16)
print("Exporting SigLIP to ONNX...")
torch.onnx.export(
    vision_tower, dummy_pixel,
    str(VISION_DIR / "siglip_vision.onnx"),
    input_names=["pixel_values"],
    output_names=["image_features"],
    dynamic_axes={"pixel_values": {0: "batch"}},
    opset_version=17,
)

# Save projector weights separately
torch.save(projector.state_dict(), str(VISION_DIR / "mm_projector.pt"))

onnx_size = (VISION_DIR / "siglip_vision.onnx").stat().st_size / 1e9
print(f"\nSigLIP ONNX: {onnx_size:.2f} GB")
print(f"Projector: {VISION_DIR / 'mm_projector.pt'}")
print("Convert to TensorRT: trtexec --onnx=siglip_vision.onnx --saveEngine=siglip.plan --fp16")

del full_model
torch.cuda.empty_cache()
````

## Troubleshooting

| Error | Cause | Fix |
|-------|-------|-----|
| `FakeTensor ... aten.detach` | Called `for_inference()` before save | Restart runtime, reload, save first |
| `Params4bit ... FakeTensor` | FakeTensorMode persists after `del model` | Full runtime restart required |
| `"not a PeftModel"` warning | Unsloth doesn't detect PeftModel wrapper | Run `model.merge_and_unload()` before save |
| `config.json does not exist` | GGUF saved to empty Drive folder | Save locally first, copy .gguf to Drive |
| `string indices must be integers` | Chat template expects list-of-dicts | Use `[{"type": "text", "text": prompt}]` not plain string |
| `Skipping ... no quant_state found` | bitsandbytes version mismatch | Use `google/gemma-3-12b-it` + `load_in_4bit=True` |
| Model outputs nonsense/math | Base model weights bad or LoRA not merged | Two-step load + `merge_and_unload()` before save |
| `attention_mask not set` | pad_token == eos_token | Pass `attention_mask` or ignore |
| Low disk for merged save | 24 GB needed for 16-bit shards | Save LoRA + GGUF instead, merge later |
| `NameError: model not defined` | `del model` was called earlier | Restart runtime, reload from checkpoint |
| TRT-LLM engine won't load on different GPU | Engine built for specific SM arch | Both GPUs must be Ampere+ for PTX JIT |
